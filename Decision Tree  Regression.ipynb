```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
```

```python
df=pd.read_csv(r"C:\Users\LENOVO\Downloads\Decision Tree Classifier.csv")
df.head()
```

```python
df["Gender"]=df["Gender"].map({"Male":0,"Female":1})
df
```

```python
df["Married"]=df["Married"].map({"No":0,"Yes":1})
df.head()
```

```python
df["Education"]=df["Education"].map({"Graduate":1,"Not Graduate":0})
df.head()
```

```python
df["Self_Employed"]=df["Self_Employed"].map({"Yes":1,"No":0})
df.head()
```

```python
df["Property_Area"]=df["Property_Area"].map({"Urban":1,"Semiurban":0})
df.head()
```

```python
df["Loan_Status"]=df["Loan_Status"].map({"Y":1,"N":0})
df.head()
```

```python
x=df[["Customer_ID","Gender","Married","Dependents","Education","Self_Employed","Applicant_Income","Coapplicant_Income","Loan_Amount","Loan_Amount_Term","Credit_History","Property_Area"]]
y=df[["Loan_Status"]]
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
model=DecisionTreeClassifier()
```

```python
model.fit(x_train,y_train)
```

```python
prediction=model.predict(x_test)
```

```python
accuracy=accuracy_score(y_test,prediction)
```
