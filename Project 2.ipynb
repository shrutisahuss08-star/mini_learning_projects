```python
import pandas as pd 
```

```python
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("Downloads\Titanic.csv")
```

```python
gender=df["Gender"].value_counts()
plt.bar(gender.index,gender.values)
plt.title("Male and Female")
plt.xlabel("Gender")
plt.ylabel("count")
plt.grid()
plt.show()
```

```python
a=df["Fare"]
plt.plot(a)
plt.title("Fare across passenger")
plt.xlabel("Passenger")
plt.ylabel("Fare")
plt.grid()
plt.show()
```

```python
p=df["Embarked"].value_counts()
plt.pie(p.values,labels=p.index)
plt.title("Passenger Boarded")
plt.xlabel("Passenger")
plt.grid()
plt.show()
```

```python
Age=df["Age"]
plt.hist(Age)
plt.title("Age")
plt.xlabel("Gender")
plt.ylabel("Age")
plt.grid()
plt.show()
```

```python
a=df["Age"]
b=df["Fare"]
plt.scatter(a,b)
plt.title("Relation Between Age & Fare")
plt.xlabel("Age")
plt.ylabel("fare")
plt.grid()
plt.show()
```

```python
pc=df["Pclass"].value_counts()
plt.barh(pc.index,pc.values)
plt.title("Passenger class")
plt.xlabel("Pclass")
plt.ylabel("count")
plt.grid()
plt.show()
```

```python
survival_count=df["Survived"].value_counts()
plt.pie(survival_count.values,labels=["Not survived","Suvived"])
plt.title("Survival Distribution")
plt.grid()
plt.show()
```
